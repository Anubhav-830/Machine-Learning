```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set()
```

Plotting the Vector

```python
plt.quiver(0,0,4,5)
plt.show()
```

```python
plt.quiver(0,0,4,5,angles='xy',scale_units='xy',scale=1)
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

```python
plt.quiver(0,0,4,5,angles='xy',scale_units='xy',scale=1 ,color='b')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

```python
plt.quiver(0,0,4,5,angles='xy',scale_units='xy',scale=1,color='b')
plt.quiver(0,0,-3,-6,angles='xy',scale_units='xy',scale=1,color='r')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Addition of two vector

```python
vector_1 = np.asarray([0,0,2,3])

vector_2 = np.asarray([0,0,3,2])

sum = vector_1 + vector_2
print(sum)
```

```python
plt.quiver(0,0,2,3,angles='xy',scale_units='xy',scale=1,color='b')
plt.quiver(0,0,3,-2,angles='xy',scale_units='xy',scale=1,color='r')
plt.quiver(0,0,5,1,angles='xy',scale_units='xy',scale=1,color='y')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Substraction of two vector

```python
vector_1 = np.asarray([0,0,2,3])

vector_2 = np.asarray([0,0,3,-2])

difference = vector_1 - vector_2
print(difference)
```

```python
plt.quiver(0,0,2,3,angles='xy',scale_units='xy',scale=1,color='b')
plt.quiver(0,0,3,-2,angles='xy',scale_units='xy',scale=1,color='r')
plt.quiver(0,0,-1,5,angles='xy',scale_units='xy',scale=1,color='y')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Multiplying the vectors

```python
vector_1 = np.asarray([0,0,2,3])

vector_2 = np.asarray([0,0,3,-2])

multiply = vector_1 *  vector_2
print(multiply)
```

```python
plt.quiver(0,0,2,3,angles='xy',scale_units='xy',scale=1,color='b')
plt.quiver(0,0,3,-2,angles='xy',scale_units='xy',scale=1,color='r')
plt.quiver(0,0,6,-6,angles='xy',scale_units='xy',scale=1,color='y')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```
