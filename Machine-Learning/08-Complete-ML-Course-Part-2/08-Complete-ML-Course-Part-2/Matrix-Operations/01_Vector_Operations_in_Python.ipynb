```python
import numpy as np
```

Dot Product Of Two Vector

```python
a = np.array([2,3])
b = np.array([4,4])

a_dot_b = np.dot(a,b)
```

```python
print(a_dot_b)
```

```python
c = np.array([4,5,6])
d = np.array([7,8,9])

c_dot_d = np.dot(c,d)
```

```python
print(c_dot_d)
```

Cross Product Of Two Vector

```python
a = np.array([4,5,7])
b = np.array([7,8,9])

a_cross_b =np.cross(a,b)
```

```python
print(a_cross_b)
```

Projection of "a" Vector on "b" Vector

```python
a = np.array([2,5])

v = np.array([8,-6])
#magnitude of v vector

magnitude_of_v = np.sqrt(sum(v**2))

proj_of_a_on_b = (np.dot(a,v)/magnitude_of_v**2)*v
```

```python
print(proj_of_a_on_b)
```
